# What is Tuple?

```python
# Definition
# Tuple is a collection usd to store multiple items in one variable.

# Tuple is ordered and immutable.

# Creating a tuple
fruits = ("apple","banana","mango")

# print tuple
print(fruits)
```

# Tuple Features

```python
# Tuple features:
# 1. Ordered
# 2. Immutable
# 3. Allows duplicate valeus

numbers= (1, 2, 2, 3)

print(numbers)
```

# Access Tuple Items

```python
# Access tuple items using index

colors = ("Red","Green","Blue")

#print first item
print(colors[0])
```

# Negative Indexing

```python
# Negative indexing starts from end
colors = ("Red","Green","Blue")

#print last item
print(colors[-1])
```

# Tuple Slicing

```python
# Slicing returns part of tuple

numbers = (10, 20, 30, 40, 50)

# Print values from index 1 to 3
print(numbers[1:4])
```

# Tuple Length

```python
# len() returns total items

fruits = ("apple", "banana", "mango")

print(len(fruits))
```

# Single Item Tuple

```python
# Single item tuple needs comma

data = ("Python",)

print(type(data))
```

# Tuple Constructor

```python
# Create tuple using tuple()

numbers = tuple((1, 2, 3))

print(numbers)
```

# Loop Through Tuple


```python
# Loop through tuple items
fruits = ("apple","banana","mango")

for fruit in fruits:

    print(fruit)
```

# Check Item Exists

```python
# check item existence

fruits = ("apple", "banana", "mango")

if "banana" in fruits:
    print("Item Found")
```

# Tuple Concatenation

```python
# Join tuples using +

a=(1,2)
b=(3,4)

# Merge tuples
print(a+b)
```

# Tuple Repetition

```python
# Repeat tuple using *

numbers = (1, 2)

print(numbers * 3)
```

# Tuple Packing

```python
# Packing means storing multiple values
student = ("Aryan",17,"Python")

print(student)
```

# Tuple Unpacking

```python
# Unpacking extracts values
student = ("Aryan", 17, "Python")

name, age, course = student

print(name)
print(age)
print(course)
```

# Using count() Method

```python
# count() counts repeated values

numbers = (1, 2, 2, 3)
print(numbers.count(2))
```

# Using index() Method

```python
#index() returns position
fruits = ("apple", "banana","mango")

print(fruits.index("banana"))
```

# Nested Tuple

```python
# Tuple inside another tuple

data = ((1, 2),(3, 4))

print(data)
```

# Tuple with List

```python
# Tuple containing list

data = ("Python",[1, 2, 3])
print(data)
```

# Convert Tuple to List

```python
# convert tuple into list

fruits = ("apple", "banana")

fruit_list = list(fruits)

print(fruit_list)
```

# Convert List to Tuple

```python
# Convert list into tuple

numbers = [1, 2, 3]

number_tuple = tuple(numbers)

print(number_tuple)
```

# Immutable Nature of Tuple

```python
# tuples cannnot be changed

fruits = ("apple", "banana")

# This gives error
# fruits[0] = "mango"

print(fruits)
```

# Tuple in Functions

```python
# Functions can return tuples

def student():
    return ("Aryan",17)
#store returned tuple
data=student()

print(data)
```

# Tuple Comprehension

```python
# Generator expression example

numbers = tuple(x*x for x in range(5))

print(numbers)
```

# Difference Between List and Tuple

```python
# List vs Tuple

# List is mutable
my_list = [1, 2, 3]

# Tuple is immutable
my_tuple = (1,2,3)

print(my_list)
print(my_tuple)
```
